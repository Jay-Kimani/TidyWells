# Tidywells quick start

Use your own downloaded Petrinex or BC Energy Regulator files. Set `FOLDER` to where you saved them.

```python
import tidywells as tw
FOLDER = 'path/to/your/files'
```

## Alberta or Saskatchewan: load monthly volumes

```python
vol = tw.read_volumetrics(FOLDER, province='AB')   # use 'SK' for Saskatchewan
vol.head()
```

## One well's monthly series and a decline fit

```python
well_id = vol['well_id'].iloc[0]
s = tw.monthly_series(vol, well_id, 'OIL')            # m3/day by month
t = [i / 12 for i in range(len(s))]
fit = tw.fit_arps(t, s.values * tw.loaders.M3_TO_BBL)
fit
```

## Forecast and NPV

Prices and costs are yours to supply; the numbers below are placeholders.

```python
f = tw.forecast(fit['qi'], fit['di'], fit['b'], years=10)    # dict: t, rate, volume
value = tw.npv(f['volume'], f['t'], price=70.0, opex_per_unit=20.0)
value
```
